```python
import pandas as pd
url = 'https://docs.google.com/spreadsheets/d/e/2PACX-1vR7rwv4mBdyx-ZgO37FOBgLCIjZuoTAnekbaX1q7pYoK1kl83tPI4xdYmJK8RG4zsjstIvs7Lj_WRh7/pub?output=csv'
raw_data = pd.read_csv(url)
df = raw_data
df.head()
```

```python
df['price_clean'] = df['price_raw'].str.replace('£', '', regex=False)
df['price_clean'] = pd.to_numeric(df['price_clean'], errors='coerce')
print(df[['product_nm', 'price_raw', 'price_clean']].head())
print("\nColumn Data Type:", df['price_clean'].dtype)
```

```python
import datetime
df['scraped_date'] = datetime.datetime.now().strftime('%Y-%m-%d')
final_dataset = df[['product_nm', 'price_clean', 'scraped_date']]
final_dataset.to_csv('clean_pricing_data.csv', index=False)
print("Export Successful!")
```
