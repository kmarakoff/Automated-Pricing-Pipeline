```python
import pandas as pd

# 1. Store your live Google Sheet link in a variable
url = 'https://docs.google.com/spreadsheets/d/e/2PACX-1vR7rwv4mBdyx-ZgO37FOBgLCIjZuoTAnekbaX1q7pYoK1kl83tPI4xdYmJK8RG4zsjstIvs7Lj_WRh7/pub?output=csv'

# 2. Tell pandas to read that web link directly as a CSV dataset
raw_data = pd.read_csv(url)

# 3. Use your shortcut to rename it to 'df' for your templates
df = raw_data

# 4. Look at the first 5 rows to make sure it loaded correctly
df.head()
```

```python
# 1. Use the .str.replace action to target the '£' symbol and erase it
df['price_clean'] = df['price_raw'].str.replace('£', '', regex=False)

# 2. Force Python to convert the column from a text "string" into a math-ready decimal number (float)
df['price_clean'] = pd.to_numeric(df['price_clean'], errors='coerce')

# 3. Look at your new column and check its mathematical data type
print(df[['product_nm', 'price_raw', 'price_clean']].head())
print("\nColumn Data Type:", df['price_clean'].dtype)
```

```python
import datetime

# 1. Create a timestamp column showing today's date
df['scraped_date'] = datetime.datetime.now().strftime('%Y-%m-%d')

# 2. Isolate just our clean columns using your 'product_nm' name
final_dataset = df[['product_nm', 'price_clean', 'scraped_date']]

# 3. Export this clean dataset to a new CSV file inside your Colab environment
final_dataset.to_csv('clean_pricing_data.csv', index=False)

print("Pipeline Export Success! Check your sidebar for 'clean_pricing_data.csv'.")
```
